```python
import numpy as np

from rl.planning import (
    Transition,
    policy_evaluation,
    policy_improvement,
    policy_iteration,
    value_iteration,
)

transition_model = {
    # State 0: starting state
    0: {
        # Action 0: move to state 1
        0: [
            Transition(
                probability=1.0,
                next_state=1,
                reward=0.0,
                terminated=False,
            ),
        ],
        # Action 1: usually move to state 2, sometimes terminate
        1: [
            Transition(
                probability=0.7,
                next_state=2,
                reward=0.0,
                terminated=False,
            ),
            Transition(
                probability=0.3,
                next_state=3,
                reward=1.0,
                terminated=True,
            ),
        ],
    },

    # State 1: high-reward but partly risky state
    1: {
        # Action 0: usually terminate with reward 5
        0: [
            Transition(
                probability=0.8,
                next_state=3,
                reward=5.0,
                terminated=True,
            ),
            Transition(
                probability=0.2,
                next_state=0,
                reward=-1.0,
                terminated=False,
            ),
        ],
        # Action 1: move to state 2 with reward 1
        1: [
            Transition(
                probability=1.0,
                next_state=2,
                reward=1.0,
                terminated=False,
            ),
        ],
    },

    # State 2: moderate reward state
    2: {
        # Action 0: terminate with reward 2
        0: [
            Transition(
                probability=1.0,
                next_state=3,
                reward=2.0,
                terminated=True,
            ),
        ],
        # Action 1: return to state 1 or terminate with reward 3
        1: [
            Transition(
                probability=0.5,
                next_state=1,
                reward=0.0,
                terminated=False,
            ),
            Transition(
                probability=0.5,
                next_state=3,
                reward=3.0,
                terminated=True,
            ),
        ],
    },

    # State 3: terminal state
    3: {
        0: [
            Transition(
                probability=1.0,
                next_state=3,
                reward=0.0,
                terminated=True,
            ),
        ],
        1: [
            Transition(
                probability=1.0,
                next_state=3,
                reward=0.0,
                terminated=True,
            ),
        ],
    },
}
```

```python
transition_model = {
    0: {
        0: [Transition(1.0, 0, 0.0, True)],
        1: [Transition(1.0, 1, 0.0, False)],
    },
    1: {
        0: [Transition(1.0, 1, 1.0, True)],
        1: [Transition(1.0, 0, 0.0, True)],
    },
}
```

### Policy Evaluation

```python
initial_policy = np.array(
    [1, 1, 0, 0],
    dtype=np.int64,
)

evaluated_values = policy_evaluation(
    policy=initial_policy,
    transition_model=transition_model,
    gamma=0.9,
    theta=1e-12,
)

print("Initial policy:", initial_policy)
print("Policy values:", evaluated_values)
```

### Policy Improvement

```python
improved_policy = policy_improvement(
    values=evaluated_values,
    transition_model=transition_model,
    gamma=0.9,
)

print("Improved policy:", improved_policy)
```

### Policy Iteration

```python
policy_result = policy_iteration(
    transition_model=transition_model,
    gamma=0.9,
    theta=1e-12,
)

print("Optimal policy:", policy_result.policy)
print("Optimal values:", policy_result.values)
print("Iterations:", policy_result.iterations)
```

### Value Iteration

```python
value_result = value_iteration(
    transition_model=transition_model,
    gamma=0.9,
    theta=1e-12,
)

print("Optimal policy:", value_result.policy)
print("Optimal values:", value_result.values)
print("Iterations:", value_result.iterations)
```
